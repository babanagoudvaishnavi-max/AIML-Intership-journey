# Student Performance Prediction & Visualization
This notebook demonstrates analysis of study hours, attendance, and final grades with combined visualizations.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Load Dataset

```python
df = pd.read_csv('../data/dataset.csv')
df.head()
```

## 2. Data Visualization (Single Frame)

```python
sns.set(style="whitegrid")

# Create ONE FRAME with 4 plots
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Correlation Heatmap
sns.heatmap(
    df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    ax=axes[0, 0]
)
axes[0, 0].set_title("Feature Correlations")

# 2. Study Hours vs Final Grade
sns.regplot(
    x='Study_Hours',
    y='Final_Grade',
    data=df,
    scatter_kws={'s': 60},
    line_kws={'linestyle': '--'},
    ax=axes[0, 1]
)
axes[0, 1].set_title("Study Hours vs. Final Grade")

# 3. Attendance vs Final Grade
sns.regplot(
    x='Attendance',
    y='Final_Grade',
    data=df,
    scatter_kws={'s': 60, 'color': 'green'},
    line_kws={'linestyle': '--', 'color': 'orange'},
    ax=axes[1, 0]
)
axes[1, 0].set_title("Attendance vs. Final Grade")

# 4. Distribution of Final Grades
sns.histplot(
    df['Final_Grade'],
    bins=8,
    kde=False,
    color='purple',
    ax=axes[1, 1]
)
axes[1, 1].set_title("Distribution of Final Grades")
axes[1, 1].set_xlabel("Final Grade")
axes[1, 1].set_ylabel("Number of Students")

# Main Title
fig.suptitle(
    "Study Hours and Attendance Strongly Drive Student Success",
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()

# Save output
plt.savefig('../screenshots/model_output_overview.png', dpi=300)

plt.show()
```
